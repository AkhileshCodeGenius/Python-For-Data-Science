# copy

#### The .copy() method in Python is used to create a shallow copy of a list or dictionary. This means it copies the references to the objects inside the list or dictionary, but not the actual objects themselves. Changes to mutable objects inside the copied collection will reflect in both the original and the copy.

```python
my_list = [1, 2, 3, 4]
copied_list = my_list.copy()
print(copied_list)  # Output: [1, 2, 3, 4]

copied_list.append(5)
print(copied_list)  # Output: [1, 2, 3, 4, 5]
print(my_list)     
```

```python
my_List = [1,2,3,4,5,6]
my_list2 = my_List.copy()
print(my_List)

my_list2.append(7)
print(my_list2)
print(my_List)
```
