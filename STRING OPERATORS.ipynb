#### . upper ()
.upper() ek method hai jo strings ke characters ko uppercase mein convert karta hai. Jab aap is method ko kisi string par call karte hain, woh string ke sabhi characters ko uppercase mein badal deta hai aur ek naya string return karta hai jismein sabhi characters uppercase hote hain.

```python
# . upper() ka mtlb Capital letter
# lower case ko upper case  me badal deta hai 
#  small letter ko capital letter  me badal deta hai 
my_string = 'hello world'
my_string.upper()
```

```python
# . upper()
my_string = 'hello world'
print(my_string.upper())
```

```python
# .lower() ka matlb small letter
# upper case ko lower case  me badal deta hai 
# capital letter ko small letter  me badal deta hai 
my_string = 'HELLO WORLD '
my_string.lower()
```

## . split ()
-  Splits the string into a list of substrings based on the separator.
-  The given argument will be removed and splitted wherver it occurs to the string.
-  The  result will come in list.
-  By default the arquement is space.

```python
# . split() Splits the string into a list of substrings based on the separator.
# The given argument will be removed and splitted wherver it occurs to the string.
# The result will come in list.
#By default the arquement is space.
my_string = 'hello world'
my_string.split(',')   
```

```python

my_string = 'hello world'
print(my_string.split('a'))
```

 Joins elements of an iterable (like a list) into a string using str as a delimiter.
